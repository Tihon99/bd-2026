# 1. Динамика соревнований
Число соревнований по годам и типам, призовые фонды, число команд, тематика по тегам.

```python
import sys
sys.path.append("..")

import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load

sns.set_theme(style="whitegrid")
comp = load("Competitions")
comp.shape
```

```python
comp["HostSegmentTitle"].value_counts()
```

Оставляем основные типы — остальные (InClass, Community, Recruitment) учебные или частные.

```python
MAIN_TYPES = ["Featured", "Research", "Playground", "Getting Started"]
comp = comp[comp["HostSegmentTitle"].isin(MAIN_TYPES)].copy()
comp["year"] = comp["EnabledDate"].dt.year

by_year = comp.groupby(["year", "HostSegmentTitle"]).size().unstack(fill_value=0)
by_year.plot(kind="bar", stacked=True, figsize=(12, 5), title="Соревнования по годам и типам")
plt.ylabel("число соревнований")
plt.savefig("../reports/figures/01_competitions_by_year.png", bbox_inches="tight")
plt.show()
```

## Призовые фонды

```python
comp["RewardType"].value_counts(dropna=False)
```

```python
usd = comp[comp["RewardType"] == "USD"]
usd.groupby("year")["RewardQuantity"].agg(["count", "median", "sum"])
```

## Число команд и участников

```python
comp.groupby("year")[["TotalTeams", "TotalCompetitors"]].median().plot(
    marker="o", figsize=(10, 4), title="Медианное число команд и участников"
)
plt.show()
```
